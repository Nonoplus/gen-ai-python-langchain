A text splitter is a utility (commonly used in frameworks like LangChain) that breaks large text into smaller chunks so it can be processed efficiently by language models.

```python
# # Text loader
from langchain_community.document_loaders import TextLoader
loader = TextLoader('speech.txt')
docs=loader.load()
docs
```

```python
from langchain_text_splitters import CharacterTextSplitter
text_slplitter = CharacterTextSplitter(separator="\n\n",chunk_size=100,chunk_overlap=20)
text_slplitter.split_documents(docs)
```

```python
speech=""
with open("speech.txt") as f:
    speech = f.read()

text_splitter = CharacterTextSplitter(chunk_size=100,chunk_overlap=20)
text = text_splitter.create_documents([speech])
print(text[0])
print(text[1])
```
